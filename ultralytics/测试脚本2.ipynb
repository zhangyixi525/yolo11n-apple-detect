```python
from ultralytics import YOLO

model = YOLO(r"G:\Projects\py_projects\yolo\ultralytics\runs\detect\train\weights\best.pt")

metrics = model.val(
    data=r"G:\Projects\py_projects\yolo\apple_dataset\data.yaml",
    split="test",
    imgsz=640,
    batch=16,
    save=True,
    plots=True,
    name="test"
)

print("="*50)
print(f"mAP50:      {metrics.box.map50:.4f}")
print(f"mAP50-95:   {metrics.box.map:.4f}")
print("-"*50)
# 分别打印两个类别
p_arr = metrics.box.p
r_arr = metrics.box.r
ap50_arr = metrics.box.ap50

print(f"bad_apple  P: {p_arr[0]:.4f}   R: {r_arr[0]:.4f}   AP50: {ap50_arr[0]:.4f}")
print(f"good_apple P: {p_arr[1]:.4f}   R: {r_arr[1]:.4f}   AP50: {ap50_arr[1]:.4f}")
print("="*50)
print(f"可视化图片、图表输出目录：{metrics.save_dir}")

# fuse 融合Conv+BN，推理时的真实状态
model.fuse()

# 打印汇总信息，返回字典可以存数据
info = model.info(detailed=False, imgsz=640)
print(info)
```
