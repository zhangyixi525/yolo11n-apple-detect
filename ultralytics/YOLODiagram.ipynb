```python
import os
import subprocess
from IPython.display import SVG, display

## ========== 配置区 ==========
YOLO_DIAGRAM_EXE = r"G:\Anaconda3\envs\pytorch\Scripts\yolo-diagram.exe"
YAML_PATH = r"G:\Projects\py_projects\yolo\ultralytics\ultralytics\cfg\models\11\yolo11n-slim.yaml"
OUTPUT_DIR = r"G:\Projects\py_projects\yolo\yolo-diagram\export_figs"
TITLE_NET = "YOLO11n-Slim 苹果检测网络结构"
THEME = "paper-mono"

MODULE_LIST = [
    # 自定义模块
    (r"G:\Projects\py_projects\yolo\ultralytics\ultralytics\nn\modules\ADown.py", "ADown"),
    (r"G:\Projects\py_projects\yolo\ultralytics\ultralytics\nn\modules\block.py", "C2PSA_TA"),
    (r"G:\Projects\py_projects\yolo\ultralytics\ultralytics\nn\modules\block.py", "BiFPN_Concat"),
    (r"G:\Projects\py_projects\yolo\ultralytics\ultralytics\nn\modules\conv.py", "GhostConv"),
    # YOLO11 官方模块
    (r"G:\Projects\py_projects\yolo\ultralytics\ultralytics\nn\modules\block.py", "C2PSA"),
    (r"G:\Projects\py_projects\yolo\ultralytics\ultralytics\nn\modules\conv.py", "Conv"),
    (r"G:\Projects\py_projects\yolo\ultralytics\ultralytics\nn\modules\block.py", "SPPF"),
]
## =====================================

os.makedirs(OUTPUT_DIR, exist_ok=True)

def run_cmd(cmd, desc):
    print(f"\n===== {desc} =====")
    print("CMD:", cmd)
    res = subprocess.run(cmd, capture_output=True, encoding="utf-8")
    print("STDOUT:\n", res.stdout)
    print("STDERR:\n", res.stderr)
    if res.returncode != 0:
        print(f"{desc} 失败，跳过，继续下一个")
        return None
    return res

# 1. 绘制网络总图
out_net = os.path.join(OUTPUT_DIR, "yolo11n_slim_arch.svg")
cmd_net = [
    YOLO_DIAGRAM_EXE,
    "network",
    YAML_PATH,
    "--output", out_net,
    "--title", TITLE_NET,
    "--theme", THEME
]
run_cmd(cmd_net, "绘制YOLO网络总图")
print(f"总图输出：{out_net}")
display(SVG(out_net))

# 2. 批量绘制所有模块（失败不中断）
for pyfile, cls_name in MODULE_LIST:
    out_mod = os.path.join(OUTPUT_DIR, f"{cls_name.lower()}.svg")
    cmd_mod = [
        YOLO_DIAGRAM_EXE,
        "module",
        "--source", pyfile,
        "--class-name", cls_name,
        "-o", out_mod,
        "--theme", THEME
    ]
    ret = run_cmd(cmd_mod, f"绘制模块 {cls_name}")
    if ret is not None:
        print(f"{cls_name} 输出：{out_mod}")
        display(SVG(out_mod))
```
