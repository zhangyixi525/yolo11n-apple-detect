```python
# 获取参数名
ckpt = YOLO("yolo11n.pt")
model_keys = set(model.model.state_dict().keys())
ckpt_keys = set(ckpt.model.state_dict().keys())

missing_in_ckpt = model_keys - ckpt_keys   # 你的模型独有：随机初始化
unused_ckpt = ckpt_keys - model_keys      # 预训练有，你模型没有：被丢弃

print(f"模型独有参数（随机初始化，共{len(missing_in_ckpt)}个）")
for k in sorted(missing_in_ckpt):
    print(k)

print(f"\n预训练权重多余参数不加载,共{len(unused_ckpt)}个）")
for k in sorted(unused_ckpt):
    print(k)
```
